```python
import pyspark
print(pyspark.__version__)  # doit afficher 3.5.0
```

```python
from pyspark.sql import SparkSession

# Connexion au cluster
spark = SparkSession.builder \
    .appName("WordCountExample") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

# Exemple de données
data = ["Bonjour le monde", "PySpark avec Docker", "MKADIA reco app", "Bonjour Spark"]

# Créer un DataFrame
df = spark.createDataFrame([(line,) for line in data], ["text"])

# Transformation : compter le nombre de mots
from pyspark.sql.functions import explode, split

word_counts = df.select(explode(split(df.text, " ")).alias("word")) \
                .groupBy("word").count()

word_counts.show()

spark.stop()
```
