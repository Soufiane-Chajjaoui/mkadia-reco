```python
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.preprocessing import TransactionEncoder
import pandas as pd

transactions = [
    ["shrimp","almonds","avocado","green tea"],
    ["burgers","meatballs","eggs"],
    ["chutney"],
    ["turkey","avocado"],
    ["mineral water","milk","energy bar","green tea"],
    ["low fat yogurt"],
    ["whole wheat pasta","french fries"]
]

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = fpgrowth(df, min_support=0.2, use_colnames=True)
print(frequent_itemsets)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("ClusterCheck") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

sc = spark.sparkContext

# Récupère les informations via statusTracker
status_tracker = sc._jsc.sc().statusTracker()
executor_infos = status_tracker.getExecutorInfos()

print("Executors connectés :")
for executor in executor_infos:
    # On utilise toString() et on parse si nécessaire
    info_str = executor.toString()
    print(info_str)

print("Nombre de partitions par défaut :", sc.defaultParallelism)

spark.stop()
```
